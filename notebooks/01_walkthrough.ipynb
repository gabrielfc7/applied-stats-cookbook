# `applied-stats-cookbook`: Statistical Inference & Resampling

A clean computational reference for applied statistical inference, hypothesis testing pipelines, multi-group ANOVA, and resampling methods with publication-ready reporting.

```python
import numpy as np
import pandas as pd
from applied_stats.parametric import one_sample_t_test, two_sample_t_test, paired_t_test
from applied_stats.nonparametric import mann_whitney_u_test, wilcoxon_signed_rank_test, kruskal_wallis_test
from applied_stats.anova import one_way_anova, two_way_anova, tukey_hsd
from applied_stats.resampling import bootstrap_ci, permutation_test
from applied_stats.reporting import PublicationReporter
```

## 1. Parametric & Non-Parametric Hypothesis Testing

```python
rng = np.random.RandomState(42)
control = rng.normal(10.0, 2.0, 35)
treatment = rng.normal(12.5, 2.8, 30)

res_welch = two_sample_t_test(treatment, control, equal_var=False)
res_mwu = mann_whitney_u_test(treatment, control)

reporter = PublicationReporter([res_welch, res_mwu])
print(reporter.to_markdown())
```

## 2. Multi-Group One-Way ANOVA & Post-Hoc Tukey HSD

```python
g1 = rng.normal(50, 4, 25)
g2 = rng.normal(55, 5, 25)
g3 = rng.normal(63, 4, 25)

anova_res = one_way_anova(g1, g2, g3)
print(anova_res.to_dataframe())

hsd_res = tukey_hsd(g1, g2, g3)
print(hsd_res)
```

## 3. Resampling Methods: Vectorized Bootstrap & Permutation Test

```python
boot = bootstrap_ci(g3, statistic_fn=np.mean, n_resamples=5000, method='bca', random_state=42)
print(boot.summary())

perm = permutation_test(g1, g3, n_permutations=5000, random_state=42)
print(perm.summary())
```
